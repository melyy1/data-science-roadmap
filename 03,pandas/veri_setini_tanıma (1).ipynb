```python
#yardimci fonskiyon:metinleri nromalize edelim.
#buyuk kucuk harf donusum,bosluklari kaldirma, noktalama isaretlerini kaldirma, turkce karakterleri duzeltme
def metin_normalize_et(series):

    return(
        series.astype(str) #veri tipini stringe donustur
        .str.strip() #bas ve sondaki bosluklari kaldir
        .str.lower() #buyuk harfleri kucuk harfe donustur
        .str.replace("ı","i") #turkce karakterleri duzeltme
        .str.replace("ş","s") #turkce karakterleri duzeltme
        .str.replace("ğ","g") #turkce karakterleri duzeltme
        .str.replace("ü","u") #turkce karakterleri duzeltme
        .str.replace("ö","o") #turkce karakterleri duzeltme
    
    )
```

```python
#6 sehir sutunundaki hatali verileri düzeltme:
df["sehir"]= metin_normalize_et(df["sehir"])
df["sehir"].unique()  #bu kodla beraber sehir saririrndaki veriler düzeldi.
```

AYKIRI DEGERLER(OUTLIERS)
Veri setini yükle
Sayısal sütunları belirle
Sayısal sütunların özet istatistiklerini hesapla
IQR ile aykırı değer sınırlarını hesapla
Aykırı değer içeren kayıtları tespit etmek
Aykırı değer sayılarını sütun bazında incelemek
Aykırı değer içeren kayıtları veri setinden temizlemek
Temizlenmiş veri setini kaydetmek

```python
import pandas as pd
#1.veri setini yuklemek
df =pd.read_csv("e_ticaret_veri_seti_tekrarlayan_kayitlar_duzenlendi.csv")

```

```python
#2. sayisal sutunların belirlenmesi:
sayisal_sutunlar= df.select_dtypes(include=["int64","float64"]).columns.tolist()
sayisal_sutunlar
```

```python
#3. sayisal sutunların özet isttaitiklerini inceleyim:
df[sayisal_sutunlar].describe()
```

```python
"""birim_fiyat
IQR= Q3(%75)-Q1(%25)
   =817-198= 619
   üst sınır= Q3+k*IQR  yani 1436
   alt sınır=Q1-k*IQR   yani -421 !! k yı genel olarak 1 alırız.
   outlier=ust ve allt sinir sisinda kalan ornekler
k=1


"""
```

```python
#4. IQR ile aykırı deger sınırlarını hesaplama
aykiri_sinirlar = {}

for sutun in sayisal_sutunlar:

    q1= df[sutun].quantile(0.25)
    q3=df[sutun].quantile(0.75)

    iqr= q3-q1
    alt_sinir= q1- 1.5*iqr
    ust_sinir= q3+ 1.5*iqr

    aykiri_sinirlar[sutun] = {
        "q1":q1,
        "q3":q3,
        "iqr":iqr,
        "alt_sinir":alt_sinir,
        "ust_sinir":ust_sinir
 }
```

```python
aykiri_sinirlar #ıqr 619 cıktı yani dogru diğerleri farklı olabilir
```

```python
#5 aykiri deger iceren kayitlari tespit et
aykiri_maskesi= pd.Series(False,index =df.index) #ilk basta tum satirlar False yani aykırı deger yok. bu yüzden maske deriz.kontorl listesi oluşturduk yani.

for sutun in sayisal_sutunlar:
    alt_sinir=aykiri_sinirlar[sutun]["alt_sinir"] #aykirisinirlar sozlugunden o sutuna ait hesaplanmıs alt sınıır cekiyor
    ust_sinir=aykiri_sinirlar[sutun]["ust_sinir"]

    sutun_maskesi= (df[sutun] <alt_sinir) | (df[sutun] > ust_sinir) #o anki sutun icin alt sınırdan kucuk mu    or   ust sınırdan buyuk mu kontrolu
    #Sadece o sütunda sınırların dışında kalan aykırı değerler için True, normal olanlar için False üreten geçici bir maske elde eder.
    aykiri_maskesi= aykiri_maskesi | sutun_maskesi #Ana maskeyi (aykiri_maskesi), o sütun için bulduğu yeni maske (sutun_maskesi) ile | (VEYA) işlemine sokup güncelliyor.
    #Mantığı: Bir satır daha önceki sütunlarda aykırı bulunmuşsa (True), bu yeni sütunda aykırı olmasa bile True kalmaya devam eder. Eğer ilk defa bu sütunda bir aykırı değer yakalandıysa, o satır da artık True olarak işaretlenir.
    #True olan satırlar: Herhangi bir sayısal sütununda en az bir tane aykırı değer (uç değer) barındıran satırlardır.
```

```python
aykiri_kayitlar= df[aykiri_maskesi]
aykiri_kayitlar  #toplam 303 tane aykiri kayıdımız var
#1.5 degerini artırsak  aykırı deger sayısı azalacak, aykiri degerlere bakarak sınırın içinde mi deigl mi diye bakmalısın
```

```python
#6. aykiri deger sayilarini sutun bazında incele
for sutun in sayisal_sutunlar:

  alt_sinir = aykiri_sinirlar[sutun]["alt_sinir"]
  ust_sinir = aykiri_sinirlar[sutun]["ust_sinir"]

  aykiri_sayi = ((df[sutun] < alt_sinir) | (df[sutun] > ust_sinir)).sum()
  print(f"{sutun}: {aykiri_sayi}")
```

```python
#7 aykırı degerler içeren kayıtları veri setinden temizleme

df_temiz= df[~aykiri_maskesi].copy() #~ bu işaret not anlamıdır maskeyi tersine çevirdik. trueyu false ,falseu true yaptık.
df_temiz.shape          #ayrıca copy yazarak bir kopya oluşturduk yapmasaydık orijinal veri bozulmasın diye hata verirdi.
```

```python
#8. temizlemiş veri setini kaydet (yeni belge olarak),
df_temiz.to_csv("e_ticaret_veri_seti_aykiri_degerler_duzenlendi.csv")
```
